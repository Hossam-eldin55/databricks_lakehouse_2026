```sql
%python
from bronze_config import INGESTION_CONFIG 
```

```sql
%python
for item in INGESTION_CONFIG:
    # Read From Sources
    print(f"Ingesting {item['source']} -> bronze.{item['target']}")
    df = (
        spark.read
        .option("header","true")
        .option("inferSchema","true")
        .csv(item['path'])
    )
    # Write in to Bronze Layer
    ( 
        df.write
        .mode("overwrite")
        .format("delta")
        .saveAsTable(f"bronze.{item["target"]}")
    )
```
